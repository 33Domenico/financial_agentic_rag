```python
import pandas as pd


df = pd.read_json("../data/raw/financebench.json",lines=True)

conteggio_valori = df['question_type'].value_counts()
print(conteggio_valori)


df_dev=df.groupby("question_type").sample(n=17,random_state=42)
conteggio_valori_dev = df_dev['question_type'].value_counts()
print(f"conteggio_valori_dev: {conteggio_valori_dev}")


df_test=df[~df["financebench_id"].isin(df_dev["financebench_id"])]
conteggio_valori_test = df_test['question_type'].value_counts()
print(f"conteggio_valori_test: {conteggio_valori_test}")


df["split"]= "test"

df.loc[df["financebench_id"].isin(df_dev["financebench_id"]),"split"]="dev"

conteggio_valori_split = df['split'].value_counts()
print(f"conteggio_valori_split: {conteggio_valori_split}")


df.to_json("../data/processed/financebench_p.json",orient="records",lines=True,force_ascii=False)

```
