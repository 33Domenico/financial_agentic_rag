```python
first_five=df.head()
doc_name_evidence_pages=[]
for row in first_five.itertuples():
    doc_name_evidence_pages.append({"doc_name": row.evidence[0]["doc_name"], "evidence_page_num": row.evidence[0]["evidence_page_num"], "evidence_text": row.evidence[0]["evidence_text"]})
doc_name_evidence_pages

```

```python
from pypdf import PdfReader
#check sulle corrispondenze veirficato

for evidence in doc_name_evidence_pages:
    reader = PdfReader(f"../data/raw/pdfs/{evidence["doc_name"]}.pdf")
    extracted_text=reader.pages[evidence["evidence_page_num"]].extract_text()
    print(extracted_text)

```

```python
from pathlib import Path
from pypdf import PdfReader

PDF_DIR = Path("../data/raw/pdfs")

pdf_paths = sorted(PDF_DIR.glob("*.pdf"))
print(f"Numero di file: {len(pdf_paths)}")

num_pages = 0
encrypted = []
failed = []

for path in pdf_paths:
    try:
        reader = PdfReader(path)
        if reader.is_encrypted:
            encrypted.append(path.stem)
        num_pages += len(reader.pages)
    except Exception as e:
        failed.append((path.stem, str(e)))

print(f"Pagine totali: {num_pages}")
print(f"PDF cifrati: {len(encrypted)}")
print(f"PDF non letti: {len(failed)}")
for name, err in failed:
    print(f"  {name}: {err}")
```
