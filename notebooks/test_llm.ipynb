```python
from dotenv import load_dotenv
import os
from langchain_openai import ChatOpenAI
```

```python
load_dotenv()


api_key = os.getenv("OPENROUTER_API_KEY") or os.getenv("OPENAI_API_KEY")
if not api_key:
    raise RuntimeError("Set OPENROUTER_API_KEY or OPENAI_API_KEY in your .env file")

model= ChatOpenAI(model="nvidia/nemotron-3.5-lightning:free", base_url="https://openrouter.ai/api/v1",api_key=api_key, temperature=0) 
```

```python
response=model.invoke("Ciao")
print(response.content)
```
