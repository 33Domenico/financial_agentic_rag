```python
from finrag.config import get_reranker

reranker = get_reranker()

print("dispositivo:", reranker.device)
print("lunghezza massima:", reranker.max_length)
print("tipo dei pesi:", next(reranker.model.parameters()).dtype)
```

```python
query = "What was the company's capital expenditure in 2018?"

docs = [
    "Purchases of property, plant and equipment (PP&E) were $1,577 million in 2018 and $1,373 million in 2017.",
    "The company's board of directors is composed of twelve members, ten of whom are independent.",
    "Capital expenditures are expected to be between $1.4 and $1.6 billion in 2019.",
]

scores = reranker.predict([(query, d) for d in docs])
for s, d in zip(scores, docs):
    print(f"{s:7.2f}  {d[:80]}")

print()
for r in reranker.rank(query, docs):
    print(r)
```
